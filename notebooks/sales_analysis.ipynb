```python
def Trend(ndf = df.copy()):
    ndf['Period'] = ndf['Order Date'].dt.to_period('M')
    ndf['month'] = ndf['Order Date'].dt.month
    sale_monthly = ndf.groupby('month',sort = True)["Sales"].sum()
    sale = ndf.groupby('Period',sort = True)["Sales"].sum()
    plt.subplots(1, 2,figsize =(14,5))
    plt.subplot(1, 2, 1)
    sale.plot(kind="line", color="orange")
    plt.subplot(1, 2, 2)
    sale_monthly.plot(kind="line")
    plt.show()
Trend(df)
```

```python
def regional_comparison(ndf = df.copy()):
    Loc = input("Enter the Region:")
    if Loc in ['South', 'West', 'Central', 'East']:
        Region = ndf[ndf['Region'] == Loc]
        data = Region[['State','Sales','Profit']].groupby("State").sum().sort_values(by = ['Profit','Sales'],ascending = False)
    else:
        data = ndf[['Region','Sales','Profit']].groupby('Region').sum().sort_values(by = ['Sales','Profit'],ascending = False)
    data.plot(kind = "bar")

regional_comparison(df)
    
```

```python
def correlation(ndf = df.copy()):
    cor = ndf[["Sales","Discount","Profit"]]
    return cor.corr()
correlation(df)
```

```python
def rfm(ndf = df.copy()):
    date_today = np.datetime64('now', 'D')
    account = ndf[["Customer ID","Order Date","Sales"]].groupby("Customer ID").agg({
        "Order Date" : lambda x: date_today - x.max(), # here in place of current date can be 
        "Customer ID": "count",
        "Sales": "sum"}).sort_values(by = ["Sales",],ascending = False)
    account.columns = ["Recency", "Frequency", "Monetary"]
    return account
rfm()
```
